### Scenario Background

Malaysia’s road network consistently ranks among the busiest and most accident‑prone in Southeast Asia. Federal roads alone account for a significant proportion of traffic incidents, particularly during peak travel periods and festive seasons, when speed limits of up to 90 km/h (and 110 km/h on expressways) are frequently exceeded in an effort to cover long distances quickly. Since 2012, the Automated Enforcement System (AES) has deployed static speed‑light and red‑light cameras at fixed points to deter speeding and dangerous cornering. However, these point‑capture devices suffer from well‑documented loopholes: drivers can simply decelerate when approaching a camera and then accelerate immediately afterward, rendering enforcement uneven and often ineffective.

To address these shortcomings, the Malaysian Government has begun rolling out the Automated Awareness Safety System (AWAS), a point‑to‑point average‑speed enforcement mechanism (Jamil et al., 2022). Figure 1 illustrates an overview of the AWAS system. AWAS leverages pairs of Ekin Spotter modular cameras equipped with 360° video surveillance and Automatic Number Plate Recognition (ANPR) to record each vehicle’s passage at two distinct checkpoints along a highway segment. By logging the exact timestamps at “Point A” and “Point B,” the system computes the travel time over a known distance (typically 1–5 km) and derives the average speed. Any average exceeding the legal limit (e.g., 110 km/h on expressways) automatically triggers a violation notice, regardless of momentary decelerations.

While AWAS promises more consistent enforcement, it also introduces significant data‑processing challenges. Each camera pair generates a continuous stream of high‑volume events—potentially thousands per minute during peak hours—that must be matched by license plate, ordered by event time, and joined across streams to compute speeds in near real time. The system must tolerate out‑of‑order or late‑arriving events (e.g., network delays), bound state growth via watermarks, and guarantee end‑to‑end exactly‑once processing to prevent duplicate violation records. These requirements make AWAS an ideal case study for a streaming Big Data architecture using Apache Kafka for ingestion, Apache Spark Structured Streaming for stateful stream–stream joins, and MongoDB for scalable storage of both raw events and flagged violations.

Reference:

Jamil, H. M., Shabadin, A., & Ibrahim, M. K. A. (2022). Automated Awareness Safety System (AwAS) for Red Light Running in Malaysia: An Analysis of Four-year Data on Its Effectiveness. Journal of the Society of Automotive Engineers Malaysia, 6(1), 19-29.

### Task 1.2 Collection Relationship

In this task, specify the relationships between collections and explain whether you choose to embed data or store references. Justify your choice in terms of:
* Read/write patterns
* Data duplication versus Join cost
* Consistency requirements (if applicable)

This section outlines the relationships between collections in the AWAS system and justifies whether referencing or embedding is used. The decisions are guided by the read/write patterns, data duplication vs. join cost, and consistency requirements for each relationship.

### Vehicle and Camera Events/Violations
- Type: One-to-Many
- A single vehicle can be involved in many camera_events and violations.

#### Implementation: Referencing via car_plate
##### Justification:
- Read/Write Pattern: Moderate read and write. Vehicle data is reused but rarely updated.
- Duplication vs Join Cost:
    - Referencing avoids duplicating static owner information across every event or violation.
    - Join cost is minimal since queries typically involve indexed car_plate.
- Consistency
    - Any updates to vehicle info (e.g., address) only require one change in the vehicles collection.

### Camera and Camera Events
- Type: One-to-Many
- Each camera generates many events.

#### Implementation: Embedding
- Read/Write Pattern: High write (streaming), high read (real-time speed checks).
- Duplication vs Join Cost:
    - Embedding of camera metadata (camera_id, speed_limit, position, longitude and latitude) ensures fast and avoid join during real-time processing
    - Memory overhead is small due to limited size of embedded data.

- Consistency:
    - Camera metadata is immutable per event — embedding captures the exact configuration at the moment of detection.

### Camers Events and Violations
- Type: One-to-One (instant) or One-to-Two (average)
- A violation is linked to one or two camera_event records.

#### Implementation: Referencing via event_id_start, event_id_end
##### Justification
- Read/Write Pattern: Both have high read and write, write for events higher than violations.

- Duplication vs Join Cost:
    - Referencing keeps violations compact and avoids duplicating full event data.
    - Only requires join during audit or violation review (not during streaming).

- Consistency:
    - Enables strong traceability — violations are linked to immutable events.

#### High Ingest Rates: Supported
camera_events is designed for high write throughput:

In the AWAS system's MongoDB data model, several deliberate trade-offs are made to balance performance, scalability, and data integrity under real-time streaming conditions. One key decision is the embedding of camera metadata (e.g., camera_id, speed_limit, position) directly within each camera_event. This introduces minor data duplication, but avoids the cost of performing a join during high-frequency writes, enabling fast, join-free stream ingestion for real-time violation detection.

Another trade-off lies in using a single violations collection for both instant and average-speed violations. While this means some fields (like camera_id_end or average_speed) will remain null in certain records, it significantly simplifies schema design, storage, and query logic — making it easier to maintain and extend.

Furthermore, vehicle data is referenced (not embedded) in both events and violations via car_plate, avoiding redundancy and allowing centralized updates if ownership details change. Although referencing introduces minor join cost during enforcement lookups, this is acceptable given the static nature and low write frequency of vehicle records.

Lastly, the system adopts a strong consistency model to ensure every query reflects the latest data, essential for timely and accurate enforcement. This choice slightly increases latency compared to eventual consistency but eliminates the risk of issuing incorrect or delayed tickets. Overall, the design favors real-time correctness, auditability, and operational efficiency, at the cost of minimal redundancy and reasonable join complexity.
